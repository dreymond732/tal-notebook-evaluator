### **Question 29 : synthèse de traitement (Correction logique) (2.0 pts)**
La fonction `analyser_phrase` doit retourner un dictionnaire contenant une liste de `mots_nettoyés` (`['le', 'tal', 'est', 'cool']`) et le `nb_caracteres` **total de la phrase originale** (avec espaces et ponctuation). Corrigez les erreurs de calcul (`nb_caracteres` et `mots_nettoyés`).

````python
phrase = "Le TAL est cool."
ponctuation_a_enlever = ".", ",", "!", "?"

def nettoyer_mot_simple(mot):
    # La fonction est correcte ici si la ponctuation est un tuple de 4 éléments.
    mot = mot.lower()
    return mot.strip(ponctuation_a_enlever)

def analyser_phrase(phrase):
    mots = phrase.split()
    mots_nettoyés = [nettoyer_mot_simple(mot) for mot in mots]

    nb_caracteres = len(mots)

    # Assurez-vous qu'il n'y a pas de chaîne vide si le split() a laissé un espace.
    mots_nettoyés = [mot for mot in mots_nettoyés if mot]

    return {
        'mots_nettoyés': mots_nettoyés,
        'nb_caracteres': nb_caracteres
    }

valeur_Q29 = analyser_phrase(phrase)
print(f"Résultat Q29 : {valeur_Q29}")
````
