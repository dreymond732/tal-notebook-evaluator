### **Question 29 : synthèse de traitement (2.0 pts)**

La fonction `nettoyer_mot_simple` est fournie : elle met un fragment en minuscules et retire les points, virgules, points d'exclamation et points d'interrogation. Conservez cette fonction.

Corrigez le calcul dans `analyser_phrase(phrase)` pour qu'elle retourne un dictionnaire contenant la liste des `mots_nettoyés` et le `nb_caracteres` **de la phrase originale**, espaces et ponctuation compris. Pour l'appel fourni, la liste doit être `['le', 'tal', 'est', 'cool']`. Le nombre de caractères doit provenir de la phrase reçue en paramètre.


````python
phrase = "Le TAL est cool."
ponctuation_a_enlever = (".", ",", "!", "?")

def nettoyer_mot_simple(mot):
    mot = mot.lower()
    for signe in ponctuation_a_enlever:
        mot = mot.replace(signe, "")
    return mot

def analyser_phrase(phrase):
    mots = phrase.split()
    mots_nettoyés = [nettoyer_mot_simple(mot) for mot in mots]

    # Corrigez la mesure du nombre de caractères.
    nb_caracteres = len(phrase)

    return {
        'mots_nettoyés': mots_nettoyés,
        'nb_caracteres': nb_caracteres
    }

valeur_Q29 = analyser_phrase(phrase)
print(f"Résultat Q29 : {valeur_Q29}")
````

### **Question 30 : rechercher le mot le plus long (2.0 pts)**

Définissez `mot_le_plus_long(mots)`, qui parcourt une liste de chaînes et retourne son élément le plus long, **sans utiliser `max`**. En cas d'égalité, conservez le premier rencontré. Une liste vide doit retourner `""`. La liste reçue ne doit pas être modifiée.

Conservez les trois appels fournis, qui vérifient une recherche ordinaire, une égalité de longueurs et une liste vide. Le résultat doit être recalculé à partir du paramètre à chaque appel.


````python
def mot_le_plus_long(mots):
    if len(mots) == 0:
        return ""
    meilleur = mots[0]
    for mot in mots:
        if len(mot) > len(meilleur):
            meilleur = mot
    return meilleur

valeur_Q30 = [
    mot_le_plus_long(["mot", "livre", "récit", "bibliothèque"]),
    mot_le_plus_long(["livre", "récit"]),
    mot_le_plus_long([]),
]
print(f"Résultat Q30 : {valeur_Q30}")
````
