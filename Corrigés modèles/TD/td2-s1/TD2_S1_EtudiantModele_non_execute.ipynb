## Exercice 1 — Nettoyer sans perdre la source — 14 min

Une chaîne est une séquence ordonnée de caractères. Sa valeur est **immuable** : une méthode de transformation produit une nouvelle chaîne. Une variable peut cependant recevoir une autre valeur par réaffectation ; conserver des noms différents permet de retrouver la source.

`strip()` retire les espaces situés aux deux extrémités d’une chaîne. Observez cet exemple distinct :

```python
etiquette = "  Atelier de lecture  "
etiquette_courte = etiquette.strip()
print(etiquette_courte)
print(len(etiquette), len(etiquette_courte))
```

Avant de l’essayer, prédisez si l’espace entre « Atelier » et « de » disparaît. Quelle variable garde la valeur initiale ?

1. À partir de `texte_brut`, construisez `texte_propre` en retirant les espaces de bord, puis affichez-le avec Q1.
2. Avec Q1b, affichez successivement les longueurs de `texte_brut` et de `texte_propre`.
3. Expliquez l’écart entre les longueurs. Les espaces entre les mots ont-ils disparu ? Pourquoi peut-on encore retrouver le texte brut ?


````python
texte_brut = "   Le TAL transforme des textes.   "
# Q1 — construisez texte_propre à partir de texte_brut.
# Ma prédiction : strip() retirera les trois espaces de chaque bord,
# mais conservera tous les espaces entre les mots.
# Dans l'exemple, l'espace entre Atelier et de reste ; etiquette garde la source.
texte_propre = texte_brut.strip()
print("Résultat Q1 :", texte_propre)
print("Résultat Q1b :", len(texte_brut), len(texte_propre))

# Mon explication raisonnée, à vérifier après exécution :
# Les longueurs attendues sont 35 et 29 : six espaces de bord ont disparu.
# strip() produit une autre chaîne ; texte_brut garde sa valeur initiale
# parce que je n'ai pas réaffecté cette variable.
````

## Exercice 4 — Passer d’une chaîne à une liste — 15 min

`split()` sans argument découpe une chaîne selon les espaces blancs et renvoie une **liste de chaînes**. Une liste est une séquence d’éléments placés dans un ordre ; ses indices commencent aussi à zéro. Plusieurs espaces successifs ne créent pas d’éléments vides avec cette forme de `split()`.

```python
annonce = "Trois  ateliers ouverts"

print(len(annonce), len(elements))
```

````python
phrase = "La traduction automatique aide parfois"
# Q4 — construisez tokens et réalisez les observations demandées.
# Ma prédiction : tokens aura cinq éléments ; son troisième sera automatique.
# Dans l'exemple, len(annonce) compte les caractères, dont les deux espaces
# successifs ; len(elements) compte trois éléments, sans élément vide.
tokens = phrase.split()
print("Résultat Q4 :", tokens, tokens[2])
print("Résultat Q4b :", len(tokens), tokens[-1])
print("Types :", type(phrase), type(tokens), type(tokens[2]))
print("Longueurs :", len(phrase), len(tokens))
print("Troisième caractère et troisième élément :", phrase[2], tokens[2])

# Mon explication : caractère, élément, type et longueur.
# phrase est une chaîne (str), tokens une liste (list), tokens[2] une chaîne (str).
# J'attends 38 caractères pour phrase et cinq éléments pour tokens.
# phrase[2] est le troisième caractère, un espace ; tokens[2] est toute
# la chaîne automatique. Un indice porte sur les unités de la séquence concernée.
# Les crochets et virgules de l'affichage représentent la structure de la liste.
````
