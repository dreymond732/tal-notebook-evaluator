### Cours — Choisir les bornes de range

`range(debut, fin, pas)` génère des entiers : début inclus, fin exclue. Sans pas explicite, le pas vaut 1. Avec un seul argument, le départ vaut 0 et cet argument donne la fin exclue. Un pas négatif permet de reculer, avec une borne finale plus petite que le début. `list(range(...))` rend les valeurs visibles. La dernière valeur souhaitée doit se trouver avant la borne exclue dans le sens du parcours.



Une autre manière de sélectionner des entiers consiste à tester leur reste de division avec `%` : `nombre % diviseur`. Un entier est pair lorsque le reste de sa division par 2 vaut 0. Cette sélection par condition et le parcours avec un pas de 2 sont deux démarches possibles ; `%` donne un reste, et non le quotient de la division.

**Exemple à observer.** Prédisez les affichages avant d’exécuter la cellule ; comparez ensuite avec votre prédiction.

```python
print("Croissant :", list(range(2, 9, 3)))
print("Décroissant :", list(range(9, 1, -2)))
for position_demo in range(3):
    print("Position :", position_demo)

# Autre donnée : reste d’une division et test de parité.
nombre_demo = 13
print("Reste par 5 :", nombre_demo % 5)
print("Pair :", nombre_demo % 2 == 0)
```

### Cours — Produire exactement vingt termes

Une boucle peut s’arrêter sur un nombre de tours plutôt que sur la valeur produite. Un compteur commence ici à 0 et augmente de 1 après chaque tour ; à la sortie, il indique le nombre de termes construits. Le terme courant et le compteur ont deux rôles différents. Préparez une table des états pour les premiers tours avant d’exécuter.

**Exemple à observer.** Prédisez les affichages avant d’exécuter la cellule ; comparez ensuite avec votre prédiction.

```python
terme_demo = 5
compteur_demo = 0
termes_demo = []
while compteur_demo < 4:
    termes_demo.append(terme_demo)
    terme_demo = terme_demo + 5
    compteur_demo = compteur_demo + 1
    print("État :", compteur_demo, terme_demo)
print("Termes :", termes_demo)
```

### Cours — Sélectionner parmi cinquante multiples

Le nombre de valeurs examinées et le nombre de valeurs retenues sont différents. Un parcours peut fabriquer chaque candidat puis n’ajouter que ceux qui satisfont un test. Le reste de division reconnaît un multiple : le reste vaut 0. Lorsque l’on parcourt les rangs 1 à N inclus, la borne finale de `range` est N + 1.

**Exemple à observer.** Prédisez les affichages avant d’exécuter la cellule ; comparez ensuite avec votre prédiction.

```python
retenus_demo = []
for rang_demo in range(1, 9):
    candidat_demo = rang_demo * 6
    if candidat_demo % 4 == 0:
        retenus_demo.append(candidat_demo)
print("Retenus :", retenus_demo)
```

```python
# Essai guidé
# Examinez les six premiers multiples positifs de 5 et retenez ceux qui sont pairs.
# Comptez à la main les candidats examinés et les valeurs gardées.
# Expliquez le rôle du rang, du candidat et du reste.

retenus_essai = []
for rang_essai in range(1, 7):
    candidat_essai = rang_essai * 5
    if candidat_essai % 2 == 0:
        retenus_essai.append(candidat_essai)
print("Essai Q5 :", retenus_essai)
# Six candidats : 5,10,15,20,25,30 ; trois retenus : 10,20,30.
# Le rang est une position, le candidat est 5 fois ce rang, le reste décide la parité.
```

### Cours — Distinguer « et » de « ou »

`condition_a and condition_b` est vraie lorsque les deux conditions sont vraies. `condition_a or condition_b` est vraie lorsqu’au moins l’une des deux l’est. Les valeurs qui satisfont les deux conditions ne doivent pas être ajoutées deux fois à une même somme. Une somme s’accumule dans une variable initialisée à 0 ; chaque valeur retenue s’ajoute à l’état précédent.

**Exemple à observer.** Prédisez les affichages avant d’exécuter la cellule ; comparez ensuite avec votre prédiction.

```python
somme_et_demo = 0
somme_ou_demo = 0
for entier_demo in range(1, 13):
    if entier_demo % 2 == 0 and entier_demo % 4 == 0:
        somme_et_demo = somme_et_demo + entier_demo
    if entier_demo % 2 == 0 or entier_demo % 4 == 0:
        somme_ou_demo = somme_ou_demo + entier_demo
print("Deux conditions :", somme_et_demo)
print("Au moins une :", somme_ou_demo)
```

```python
# Essai guidé
# Entre 1 et 10 inclus, listez les nombres divisibles par 2 et par 5, puis par 2 ou par 5.
# Calculez les deux sommes à la main, puis vérifiez-les par parcours.
# Expliquez pourquoi 10 est ajouté une seule fois dans la seconde somme.

valeurs_et_essai = []
valeurs_ou_essai = []
somme_et_essai = 0
somme_ou_essai = 0
for entier_essai in range(1, 11):
    if entier_essai % 2 == 0 and entier_essai % 5 == 0:
        valeurs_et_essai.append(entier_essai)
        somme_et_essai = somme_et_essai + entier_essai
    if entier_essai % 2 == 0 or entier_essai % 5 == 0:
        valeurs_ou_essai.append(entier_essai)
        somme_ou_essai = somme_ou_essai + entier_essai
print("Essai Q6 :", valeurs_et_essai, somme_et_essai, valeurs_ou_essai, somme_ou_essai)
# Prédiction : [10], 10 ; [2,4,5,6,8,10], 35. 10 n’est ajouté qu’une fois par condition.
```

## Q6 — Distinguer « et » de « ou »

Problème autonome : entre les bornes `a` et `b`, toutes deux incluses, calculez `somme_et`, la somme des entiers divisibles par 3 **et** par 5, puis `somme_ou`, celle des entiers divisibles par 3 **ou** par 5. Utilisez un parcours `for`. Affichez les deux sommes avec Q6.

**Pour vérifier :** écrivez dans un essai les listes des valeurs retenues ; chaque multiple commun doit contribuer une seule fois à chacune des sommes auxquelles il appartient.

**Appui documentaire.** Gérard Swinnen, *Apprendre à programmer avec Python*, exemplaire fourni (copyright 2000–2012), exercice 6.8, p. 58 : additionner entre deux bornes les multiples de 3 et de 5, puis de 3 ou de 5 ; bornes fixées dans la cellule, sans saisie interactive.

```python
a = 1
b = 30
somme_et = 0
somme_ou = 0
for entier in range(a, b + 1):
    if entier % 3 == 0 and entier % 5 == 0:
        somme_et = somme_et + entier
    if entier % 3 == 0 or entier % 5 == 0:
        somme_ou = somme_ou + entier
print("Résultat Q6 :", somme_et, somme_ou)
# Prédiction : 45 et 225. « et » retient 15 et 30 ; « ou » retient l’une ou l’autre divisibilité.
# Un seul ajout est écrit dans chaque condition : 15 et 30 ne sont pas doublés dans somme_ou.
```
