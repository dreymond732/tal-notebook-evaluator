## Exercice 22 : calculer une moyenne et un écart-type


### Repères pour Q22

On décrit ici toute la population fournie : la variance est la moyenne des carrés des écarts à la moyenne, donc divisée par n. L'écart-type est sa racine carrée et retrouve l'unité de départ. math.sqrt est autorisé dans cette question.

Prédisez les affichages du petit exemple avant de l'exécuter. Ses données sont distinctes de la question.


```python
mesures = [1, 3]
centre = sum(mesures) / len(mesures)
ecart_carre = [(v - centre) ** 2 for v in mesures]
print(centre, ecart_carre)
```

```python
# Essai personnel sur les données distinctes ci-dessus.
# Terminez à la main la variance et l'écart-type de cet exemple. Que deviennent-ils si les deux valeurs sont identiques ? Comparez population et estimation sur échantillon sans changer la convention demandée.
# Avant exécution : votre prédiction.
# Écrivez votre essai puis expliquez tout écart observé.
```

```python
valeurs = [2, 4, 4, 4, 5, 5, 7, 9]

# CORRECTION REQUISE (variance population : division par n)
moyenne = sum(valeurs) / len(valeurs)
variance = sum((x - moyenne)**2 for x in valeurs) / (len(valeurs) - 1)
ecart_type = math.sqrt(variance)

resultat_Q22 = {'moyenne': moyenne, 'ecart_type': ecart_type}
print("Résultat Q22 :", resultat_Q22)
```

### Repères pour Q24

Une recherche linéaire examine les valeurs successives. On peut conclure dès qu'une cible est trouvée ; conclure à son absence demande d'avoir terminé le parcours. L'interdiction de in concerne le test d'appartenance, pas la syntaxe for … in ….

```python
# Essai personnel sur les données distinctes ci-dessus.
# Tracez une recherche de 6 puis de 5 dans cette liste. À quelle étape chaque conclusion est-elle justifiée ? Testez aussi une liste vide après rédaction.
# Avant exécution : votre prédiction.
# Écrivez votre essai puis expliquez tout écart observé.
```

```python
def contient_liste(valeurs, cible):
    """Retourne True si cible est dans valeurs, sinon False."""
    # CORRECTION REQUISE (ne pas utiliser 'in')
    for v in valeurs:
        if v == cible:
            return False
    return True

valeur_Q24 = contient_liste([1, 4, 9], 4)
print("Résultat Q24 :", valeur_Q24)
```

## Exercice 25 : écrire une fonction récursive simple


### Repères pour Q25

Une fonction récursive s'appelle sur un problème plus petit. Un cas de base arrête les appels ; chaque chemin doit y arriver. Pour la somme demandée, le domaine est n≥1 : ne cherchez pas à traiter des entrées négatives avant ce contrat.

Prédisez les affichages du petit exemple avant de l'exécuter. Ses données sont distinctes de la question.


```python
def longueur_recursive(texte):
    if texte == "":
        return 0
    return 1 + longueur_recursive(texte[1:])
print(longueur_recursive("mot"))
```

```python
# Essai personnel sur les données distinctes ci-dessus.
# Dessinez la pile des appels pour « ab ». Expliquez ce qui diminue et la valeur renvoyée au cas de base, puis appliquez ce raisonnement à la somme, sans recopier la fonction.
# Avant exécution : votre prédiction.
# Écrivez votre essai puis expliquez tout écart observé.
```

```python
def somme_n(n):
    """Retourne 1+2+...+n (n>=1)."""
    # CORRECTION REQUISE
    if n == 1:
        return 1
    return n + somme_n(n-2)

valeur_Q25 = somme_n(5)
print("Résultat Q25 :", valeur_Q25)
```
