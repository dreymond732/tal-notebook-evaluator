## Rappel — lire les annotations d'un token

Parcourir un `Doc` avec une boucle `for` permet d'examiner ses tokens l'un après l'autre.

| Attribut | Information lue |
|---|---|
| `token.text` | La forme telle qu'elle apparaît dans le texte. |
| `token.lemma_` | Le lemme prédit, c'est-à-dire une forme de référence : par exemple un singulier ou un infinitif. |
| `token.pos_` | La catégorie grammaticale prédite, notée par un label comme `NOUN` (nom), `VERB` (verbe) ou `ADJ` (adjectif). |

Un **triple** rassemble trois valeurs. Dans `(forme, lemme, categorie)`, leur place indique ce que représente chacune. Une liste de triples conserve ces trois informations pour chaque token, dans l'ordre de lecture et avec les répétitions éventuelles.

### Exemple fourni — une ligne par token

Dans l'exemple suivant, `ligne_demo` est un tuple de trois valeurs ; `append()` l'ajoute à la fin de `lignes_demo`. L'ajout a lieu pour chaque token, y compris la ponctuation.


```python
doc_demo = nlp("La carte change.")
lignes_demo = []
for token_demo in doc_demo:
    ligne_demo = (token_demo.text, token_demo.lemma_, token_demo.pos_)
    lignes_demo.append(ligne_demo)
print(lignes_demo)
```

## Rappel — sélectionner selon une catégorie

Lire un attribut et sélectionner un token sont deux opérations différentes. La condition `if` permet de conserver seulement les tokens dont la catégorie correspond au critère choisi. Dans le test `token.pos_ == "ADJ"`, `==` compare la catégorie au label des adjectifs ; il n'affecte pas une nouvelle valeur au token.

### Exemple fourni — conserver les lemmes des adjectifs

L'exemple ci-dessous travaille sur une autre phrase et sélectionne les adjectifs. L'ajout à la liste est placé **dans la condition** : les autres tokens ne sont pas retenus. La boucle garde leur ordre et ne supprime pas les répétitions.


```python
doc_adjectifs_demo = nlp("Une petite carte ancienne change.")
adjectifs_demo = []
for token_demo in doc_adjectifs_demo:
    if token_demo.pos_ == "ADJ":
        adjectifs_demo.append(token_demo.lemma_)
print(adjectifs_demo)
```
