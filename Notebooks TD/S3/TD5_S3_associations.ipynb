### Deux proportions, deux populations

La proportion d'association sachant le pivot vaut `cooc / marge_pivot`. La proportion de base de l'associé vaut `marge_associe / N`. Elles n'ont pas le même dénominateur. Ici, toutes les valeurs concernent des **présences par phrase**.

**Exemple commenté.** Sur 20 phrases, 5 contiennent « musée », 10 contiennent « ville » et 4 les deux. La proportion de « ville » parmi les phrases avec « musée » est 4/5 ; sa proportion de base est 10/20. Ce sont des descriptions du corpus, pas la probabilité de vérité d'une affirmation.


```python
# Exemple : 20 phrases, dont 5 avec musée, 10 avec ville et 4 avec les deux.
conditionnelle_exemple = 4 / 5
base_exemple = 10 / 20
print("Ville parmi les phrases avec musée :", conditionnelle_exemple)
print("Ville parmi toutes les phrases :", base_exemple)
```

### Distinguer zéro et indéfini

Une absence d'associé parmi des phrases contenant un pivot conduit à une proportion nulle. Une absence de pivot empêche de définir cette proportion : en Python, utilisez `None`, qui devient `null` dans JSON.

**Exemple commenté.** Sur dix phrases sans « volcan », on ne peut pas estimer dans ce corpus la proportion de « lave » parmi les phrases contenant « volcan ». Ajouter artificiellement 1 au dénominateur produirait une mesure différente.

L’exemple d’affichage conserve `None` sans tenter de l’arrondir. Dans JSON, cette absence de valeur s’écrit `null` ; un quotient nul reste un nombre.


```python
# Exemple : afficher un quotient ou conserver son absence.
import json
for valeur_exemple in [None, 0.0, 2 / 7]:
    if valeur_exemple is None:
        affichage_exemple = None
    else:
        affichage_exemple = round(valeur_exemple, 6)
    print(json.dumps({"proportion": affichage_exemple}))
```
