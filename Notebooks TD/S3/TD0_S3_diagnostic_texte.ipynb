## Exemple guidé — Construire un dictionnaire de fréquences

Une fréquence associe une forme à son nombre d’occurrences. La méthode `.get(cle, 0)` évite de traiter séparément le premier passage d’une forme.

```python
frequences = {}
for element in ["a", "b", "a"]:
    frequences[element] = frequences.get(element, 0) + 1
print(frequences)  # {'a': 2, 'b': 1}
```
