### Rappel — Compter et comparer des fréquences

`Counter` associe chaque élément à son nombre d'occurrences. Sa méthode `.most_common(2)` ne montre que les deux effectifs les plus élevés ; elle ne supprime rien du compte complet. `sum(compte.values())` donne le nombre total d'occurrences comptées. Dans une fonction, `return` renvoie ce résultat pour qu'un autre calcul puisse le réutiliser ; `print` ne fait que l'afficher.

Une fréquence rapportée à 1 000 unités vaut `1000 * effectif / total`, si le total est non nul. Il faut annoncer ce total : tous les tokens, les mots alphabétiques ou seulement les noms ? Deux tableaux utilisant des populations différentes ne répondent pas à la même question.

**Exemple fourni.** Exécutez la cellule suivante. Elle compte trois noms de documents ; le dénominateur 3 désigne ici toutes les occurrences de cette liste.


```python
compte_demo = Counter(["carte", "ville", "carte"])
print("Classement :", compte_demo.most_common(2))
print("Total :", sum(compte_demo.values()))
print("Occurrences de carte pour 1 000 éléments :", 1000 * compte_demo["carte"] / 3)
```

### Cours — Une expression n'est pas un lemme

Une **forme** est l'écriture d'un token ; un **lemme** est l'annotation proposée pour ce token ; une **expression** réunit plusieurs tokens dans un ordre donné. Chercher une expression demande de vérifier que ses éléments se suivent. Une famille lexicale, comme « intelligence » et « intelligent », correspond encore à un autre regroupement : elle ne découle pas automatiquement de la lemmatisation.

spaCy propose `PhraseMatcher` pour rechercher des suites de tokens. On lui fournit le vocabulaire de `nlp`, puis un motif créé par `nlp.make_doc(...)`. Avec `attr="LOWER"`, la comparaison porte sur les formes en minuscules : elle ignore la casse, mais ne remplace pas les formes par leurs lemmes. Chaque correspondance contient un identifiant de règle et deux **indices de tokens** ; la tranche `doc[debut:fin]` permet d'en lire le passage.

**Exemple fourni.** Exécutez la cellule suivante et observez le passage trouvé. Le motif reconnaît « carte ancienne », mais ne rend pas automatiquement « cartes anciennes » équivalent. Vous adapterez cet outil au TD3 ; ici, il s'agit de comprendre quel objet il recherche.


```python
from spacy.matcher import PhraseMatcher
regles_demo = PhraseMatcher(nlp.vocab, attr="LOWER")
regles_demo.add("EXPRESSION", [nlp.make_doc("carte ancienne")])
doc_demo_expression = nlp("Cette carte ancienne accompagne le catalogue.")
for identifiant_demo, debut_demo, fin_demo in regles_demo(doc_demo_expression):
    print(doc_demo_expression[debut_demo:fin_demo].text)
```

### Cours — Deux représentations du même comptage

`WordCloud` produit un **nuage de mots** à partir de fréquences. Sa méthode `.generate_from_frequencies(...)` reçoit le compte déjà construit : elle ne doit pas recalculer un texte filtré différemment. La taille visuelle suggère une importance relative ; elle ne suffit pas à lire précisément un effectif.

Matplotlib, appelé ici `plt`, permet d'afficher le nuage et de construire un graphique en barres. `plt.figure(...)` ouvre une nouvelle figure ; `imshow` affiche une image ; `axis("off")` masque ses axes ; `bar` associe une barre à chaque valeur. `xlabel`, `ylabel` et `title` précisent ce qui est représenté ; `show` affiche la figure.

**Exemple fourni.** Exécutez les deux représentations ci-dessous. Elles utilisent exactement les mêmes effectifs. Les barres et leur axe vertical permettent une lecture numérique plus précise. `random_state=42` stabilise le placement des mots dans le nuage, pas les comptes.


```python
effectifs_demo = {"carte": 4, "atlas": 2, "plan": 1}
nuage_demo = WordCloud(width=600, height=250, background_color="white", random_state=42)
nuage_demo.generate_from_frequencies(effectifs_demo)
plt.figure(figsize=(6, 2.5))
plt.imshow(nuage_demo, interpolation="bilinear")
plt.axis("off")
plt.show()

plt.figure(figsize=(6, 3))
plt.bar(list(effectifs_demo.keys()), list(effectifs_demo.values()))
plt.xlabel("Forme")
plt.ylabel("Occurrences")
plt.title("Effectifs dans le corpus de démonstration")
plt.show()
```
