### Exemple commenté — Une expression n’est pas un lemme

```python
from spacy.matcher import PhraseMatcher
regles = PhraseMatcher(nlp.vocab, attr='LOWER')
regles.add('EXPRESSION', [nlp.make_doc('carte ancienne')])
mini = nlp('Cette carte ancienne accompagne le catalogue.')
print([mini[debut:fin].text for _, debut, fin in regles(mini)])
```
`PhraseMatcher` recherche ici une suite de formes sans tenir compte de la casse. Il ne rend pas automatiquement « cartes anciennes » équivalent à « carte ancienne ». Une famille lexicale comme intelligence/intelligent est encore un autre choix.
