### Lire une dispersion

Une **dispersion** place les occurrences le long du texte. Elle montre des concentrations ou des absences, sans expliquer à elle seule leur sens. Une position relative est ici `i / N`, où `i` est un indice à partir de zéro et `N` le nombre de tokens du flux ; la dernière position reste donc inférieure à 1.

**Exemple fourni.** Dans un flux de 10 tokens, les occurrences situées aux indices 2 et 8 apparaissent à 0,2 et 0,8. Elles sont espacées dans le texte ; leur fréquence seule ne le montre pas.

Dans la cellule suivante, `eventplot` dessine un trait pour chaque position. Chaque liste intérieure représente une ligne de dispersion ; `yticks` associe cette ligne à un nom. `xlim(0, 1)` garde la même échelle relative pour toutes les lignes.


```python
plt.figure()
positions_exemple = [0.2, 0.8]
plt.eventplot([positions_exemple], colors=["navy"])
plt.xlim(0, 1)
plt.xlabel("Position relative dans le flux")
plt.yticks([0], ["exemple : forêt"])
plt.show()
```

### Lire une carte de chaleur

Une carte de chaleur doit afficher une mesure commune à toutes ses cases. Une couleur foncée ne signifie « plus d'occurrences » que si la valeur représentée est un effectif ; pour un taux, elle signifie une densité plus élevée.

**Exemple fourni.** Deux termes et deux segments donnent une matrice où chaque ligne désigne un terme, chaque colonne un segment. `imshow` représente les nombres ; il ne choisit ni l'unité ni les dénominateurs à votre place.

`ax.imshow` associe une couleur à chaque valeur de la matrice. `aspect="auto"` adapte la largeur des cases à la figure ; `cmap="Blues"` choisit la palette. `set_xticks` et `set_yticks` associent les indices aux noms des colonnes et des lignes. `fig.colorbar` ajoute la correspondance entre couleurs et nombres, avec leur unité. Les valeurs de l’exemple sont des taux déjà calculés.


```python
matrice_exemple = [[10, 30], [20, 0]]
fig, ax = plt.subplots()
im = ax.imshow(matrice_exemple, aspect="auto", cmap="Blues")
ax.set_xticks([0, 1], ["début", "fin"])
ax.set_yticks([0, 1], ["mer", "montagne"])
fig.colorbar(im, ax=ax, label="Occurrences pour mille mots")
plt.show()
```
