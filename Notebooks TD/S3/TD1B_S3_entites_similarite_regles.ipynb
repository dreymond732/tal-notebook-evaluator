## 1. Reconnaître une entité nommée

Une entité nommée est ici une mention que l'on délimite dans le texte et que l'on classe : une personne, une organisation ou un lieu, par exemple. Une mention peut contenir plusieurs tokens. Le modèle français propose les labels `PER` (personne), `ORG` (organisation), `LOC` (lieu) et `MISC` (autres catégories d'entités).

La reconnaissance comporte donc deux décisions : **où la mention commence et finit**, puis **quelle catégorie lui attribuer**. Ce sont des prédictions, pas des certitudes. Un nom propre n'est pas nécessairement détecté ; une expression reconnue peut être mal délimitée ou mal classée. Les catégories disponibles dépendent du modèle : celui-ci ne propose pas de label spécialisé `DATE` ou `MONEY` pour les dates ou les montants.

`doc.ents` contient les mentions détectées, dans l'ordre du texte. Chaque mention est un objet `Span`, un segment constitué de tokens consécutifs. Ses attributs `text` et `label_` donnent la forme du segment et son label lisible.

### Exemple fourni — lire les prédictions

Le code ci-dessous affiche chaque mention détectée sur une ligne. Si une expression que vous reconnaissez n'apparaît pas, cela signifie qu'elle n'a pas été retenue comme entité par le modèle ; l'affichage ne la supprime pas.


```python
texte_exemple_entites = "Marie Curie a étudié à Paris."
doc_exemple_entites = nlp_entites(texte_exemple_entites)
for entite in doc_exemple_entites.ents:
    print(entite.text, "→", entite.label_)
```

## 2. Extraire seulement les informations utiles

Détecter des entités et sélectionner certaines catégories sont deux opérations différentes. La détection propose des segments et leurs labels ; l'extraction conserve ensuite les segments utiles à une question de recherche.

Un **faux positif** est une mention détectée comme une entité alors que votre lecture ne justifie pas cette détection ou cette catégorie. À l’inverse, une mention pertinente peut ne pas être détectée.

Le filtre ne corrige pas le modèle. S'il n'a pas reconnu une personne, un filtre sur `PER` ne la retrouvera pas. Le label d'entité `PER` ne désigne pas une catégorie grammaticale : contrairement à `pos_`, il concerne une mention qui peut contenir plusieurs mots.

### Exemple fourni — sélectionner les lieux

L'exemple conserve uniquement les entités étiquetées `LOC`. La liste peut être vide si le modèle ne propose aucun lieu. Une fois le filtre appliqué, les personnes éventuellement reconnues ne figurent plus dans cette liste.


```python
doc_exemple_lieux = nlp_entites("Victor Hugo a vécu à Paris puis à Guernesey.")
lieux_exemple = []
for entite in doc_exemple_lieux.ents:
    if entite.label_ == "LOC":
        lieux_exemple.append((entite.text, entite.label_))
print(lieux_exemple)
```

## 3. Comparer la proximité de mots et de phrases

Un **vecteur de mot** est une liste de nombres qui représente ce mot dans un espace construit à partir de textes. Vous n'avez pas à calculer ces nombres : le modèle `md` en fournit. Deux représentations proches peuvent correspondre à des mots employés dans des contextes voisins.

La méthode `.similarity()` calcule un score de proximité entre deux objets spaCy. Pour les documents utilisés ici, spaCy compare des représentations obtenues par moyenne des vecteurs des tokens. Cette moyenne peut rapprocher des phrases qui parlent du même sujet tout en affirmant des choses différentes. L'ordre des mots et la négation ne sont pas interprétés comme le ferait un lecteur.

Le score n'est **ni un pourcentage de vérité, ni une probabilité que deux phrases disent la même chose**. Un score plus grand indique seulement une plus grande proximité selon cette représentation. Il faut interpréter la comparaison à partir des textes et de la méthode utilisée.

### Exemple fourni — calculer un score sur deux mots

Le pipeline `nlp_sim` est chargé séparément pour ces comparaisons. L'attribut `has_vector` indique si une représentation vectorielle est disponible. Une représentation présente ne garantit pas qu'elle soit utile pour tous les sens du mot. Si spaCy signale un vecteur vide ou absent, conservez cet avertissement et ne tirez pas de conclusion du score concerné.


```python
nlp_sim = spacy.load("fr_core_news_md")
mot_exemple_a = nlp_sim("rivière")
mot_exemple_b = nlp_sim("fleuve")
print("Vecteurs présents :", mot_exemple_a.has_vector, mot_exemple_b.has_vector)
print("rivière / fleuve :", mot_exemple_a.similarity(mot_exemple_b))
```

## 4. Ajouter une règle sans entraîner un nouveau modèle

Vous pouvez vouloir reconnaître une expression précise, même si le modèle ne la repère pas. `EntityRuler` est un composant de spaCy qui ajoute des entités à partir de règles. Avec une règle exacte, vous associez une expression à un label choisi.

Une règle est ici un dictionnaire comportant deux clés : `"label"` pour la catégorie, et `"pattern"` pour l'expression à reconnaître. La méthode `add_patterns()` reçoit une **liste** de ces dictionnaires. Les règles employées dans cette séance comparent des formes exactes : une autre graphie peut ne pas correspondre. Elles ne permettent pas à elles seules de comprendre le sens d'une expression dans tous ses contextes.

`add_pipe("entity_ruler", before="ner")` place les règles avant le composant de reconnaissance statistique `ner`. Les segments reconnus par le composant de règles sont alors pris en compte par `ner`. Ce placement ne réentraîne pas le modèle. Les règles peuvent imposer une annotation inadéquate si leur expression ou leur catégorie est mal choisie.

### Exemple fourni — observer le même texte avant et après une règle

L'exemple associe l'expression fictive « Atelier des Lucioles » à `ORG`. Il charge son propre pipeline à chaque exécution : vous pouvez donc relancer la cellule sans ajouter deux fois le même composant. Les autres pipelines restent disponibles pour vos comparaisons.


```python
nlp_demo_regles = spacy.load("fr_core_news_md")
texte_exemple_regle = "L'Atelier des Lucioles accueille une exposition."
doc_exemple_avant = nlp_demo_regles(texte_exemple_regle)
print("Avant la règle :")
for entite in doc_exemple_avant.ents:
    print(entite.text, entite.label_)

regles_exemple = nlp_demo_regles.add_pipe("entity_ruler", before="ner")
regles_exemple.add_patterns([
    {"label": "ORG", "pattern": "Atelier des Lucioles"}
])
print("Étapes du pipeline :", nlp_demo_regles.pipe_names)
doc_exemple_apres = nlp_demo_regles(texte_exemple_regle)
print("Après la règle :")
for entite in doc_exemple_apres.ents:
    print(entite.text, entite.label_)
```

### Exemple fourni — conserver une comparaison

Pour comparer plusieurs essais, vous pouvez conserver un dictionnaire par texte. Chaque dictionnaire contient le texte et les mentions observées avec et sans règles. Une liste extérieure rassemble ces dictionnaires.

La cellule suivante conserve les résultats de l'exemple précédent. Elle ne prédit pas ce que le modèle devrait répondre : elle lit les deux analyses déjà calculées.


```python
mentions_avant_demo = []
for entite in doc_exemple_avant.ents:
    mentions_avant_demo.append((entite.text, entite.label_))
mentions_apres_demo = []
for entite in doc_exemple_apres.ents:
    mentions_apres_demo.append((entite.text, entite.label_))
essais_demo = [{
    "texte": texte_exemple_regle,
    "sans_regles": mentions_avant_demo,
    "avec_regles": mentions_apres_demo,
}]
print(essais_demo)
```

Les deux analyses portent sur **le même texte**.
