### Présence et nombre d’occurrences

Une **marge** est le nombre de contextes contenant un terme. Dans une phrase, cinq répétitions du même terme comptent pour une seule présence. Le nombre total `N` inclut également les contextes où le pivot est absent.

**Exemple commenté.** Dans `[["mer", "mer", "port"], ["port"], ["vent"]]`, `set` enlève les répétitions à l'intérieur de chaque contexte. La marge de « mer » est 1, celle de « port » est 2, et il y a 3 contextes. Le test `in` vérifie une présence : il répond par `True` ou `False`.


```python
# Exemple : répétition d’un mot et présence dans un contexte.
contexte_exemple = ["mer", "mer", "port"]
occurrences_exemple = 0
for mot_exemple in contexte_exemple:
    if mot_exemple == "mer":
        occurrences_exemple += 1
print("Occurrences de mer :", occurrences_exemple)
print("Mer présente :", "mer" in set(contexte_exemple))
print("Formes distinctes :", len(set(contexte_exemple)))
```

### Repère — Indice token et position caractère

Sur une tokenisation illustrative de « La mer. », les tokens `La`, `mer`, `.` peuvent être décrits ainsi :

| Token | Indice dans le flux | Début en caractères |
|---|---:|---:|
| La | 0 | 0 |
| mer | 1 | 3 |
| . | 2 | 6 |

`token.i` sert à la distance entre tokens ; `token.idx` permet de retrouver les caractères du texte. Ces unités ne se soustraient pas entre elles. Sur `['rose', ',', 'bleue']`, enlever la virgule transforme la distance de 2 en 1. Gardez donc le flux non filtré de l'exercice. Pour refuser une fenêtre invalide, reprenez le mécanisme `raise ValueError` travaillé au TD3.


### Mesurer la distance dans un flux de tokens

Une fenêtre requiert un flux et une distance. Ici, une paire est constituée d'une position du pivot et d'une position de l'associé, avec **`0 < abs(i-j) <= k`**. Chaque paire de positions distinctes compte une fois. Les termes de nos essais sont distincts.

**Exemple commenté.** Dans `["soleil", "et", "la", "lune"]`, les positions sont 0 et 3 : distance 3. Retirer « et » et « la » avant de mesurer donnerait une distance de 1 et changerait la question.

Dans l’exemple exécutable, `enumerate` associe chaque élément à son indice, à partir de 0. `abs` donne une distance positive, quel que soit l’ordre de la soustraction.


```python
# Exemple : les positions appartiennent au flux complet.
mots_exemple = ["soleil", "et", "la", "lune"]
positions_exemple = []
for i, mot in enumerate(mots_exemple):
    if mot == "lune":
        positions_exemple.append(i)
print("Positions de lune :", positions_exemple)
print("Distance entre les extrémités :", abs(0 - 3))
```

### Forme, lemme et expression

La lemmatisation élargit parfois les regroupements, sans garantir leur pertinence. Une famille conceptuelle, une forme et un lemme sont trois objets différents.

**Exemple commenté.** Les formes « écoles » et « école » peuvent partager le lemme « école ». Le mot « scolaire » n'est pas ajouté pour autant. Un ensemble thématique doit être justifié séparément.

L’exemple ci-dessous montre comment choisir le champ d’un dictionnaire à lire. Les annotations y sont écrites à la main pour distinguer la lecture des données de la prédiction d’un modèle.

**Convention des expressions :** ignorez les tokens constitués uniquement d’espaces (`token.is_space`) pour rapprocher les mots d’une expression coupée par un retour de ligne ; conservez la ponctuation et les indices `token.i` et `token.idx` d’origine. Ainsi, « bon » puis un retour de ligne puis « sens » peut correspondre à « bon sens », mais « bon, sens » reste différent.
