### Cours — Lire une concordance

Une **concordance** présente une occurrence recherchée, appelée **pivot**, entourée d'un contexte gauche et d'un contexte droit. Elle permet de lire l'emploi d'un mot sans perdre sa place dans le texte.

Dans l'exemple suivant, le contexte contient au plus cinq caractères à gauche et dix à droite. `max(0, ...)` empêche le début de passer avant le texte ; `min(len(source), ...)` borne la fin. La taille du contexte sert ici à lire : ce n'est pas encore une définition de cooccurrence en mots.

**Exemple fourni.** Exécutez la cellule et comparez le pivot seul au passage qui l'entoure.


```python
source_demo_contexte = "Une carte ancienne rejoint la collection."
debut_demo_contexte = source_demo_contexte.find("carte")
if debut_demo_contexte != -1:
    fin_demo_contexte = debut_demo_contexte + len("carte")
    print("Pivot :", source_demo_contexte[debut_demo_contexte:fin_demo_contexte])
    print("Avec contexte :", source_demo_contexte[max(0, debut_demo_contexte-5):min(len(source_demo_contexte), fin_demo_contexte+10)])
```

### Cours — Reprendre une recherche et distinguer absence et erreur

Le second argument de `find(motif, depart)` indique où reprendre la recherche. Après une occurrence, avancer de la longueur du motif évite de retrouver toujours la même occurrence. Cela définit aussi une recherche **sans chevauchement** : une occurrence suivante ne commence pas à l'intérieur de la précédente. Une boucle `while` peut poursuivre la recherche tant qu'une occurrence est trouvée ; son point de départ doit avancer à chaque tour.

Un motif absent peut produire une liste vide. Un motif vide pose un autre problème : sa longueur n'aide pas à avancer. Vous pouvez le refuser explicitement avec `raise ValueError(...)`. Un bloc `try` suivi de `except ValueError` permet ensuite de vérifier ce refus sans arrêter tout le notebook. Un `except` sans type masquerait aussi d'autres erreurs.

**Deux exemples fournis.** Le premier reprend une recherche une seule fois ; il ne fournit pas la fonction de concordance. Le second refuse une largeur négative, sur un cas distinct de l'exercice. Avant de l'exécuter, prédisez le chemin suivi pour une largeur de 2.


```python
phrase_demo_reprise = "Une carte, puis une autre carte."
premier_demo = phrase_demo_reprise.find("carte")
second_demo = phrase_demo_reprise.find("carte", premier_demo + len("carte"))
print(premier_demo, second_demo)
```

```python
def verifier_largeur(largeur):
    if largeur < 0:
        raise ValueError("La largeur doit être positive ou nulle")
    return largeur

print("Entrée valide :", verifier_largeur(2))
try:
    verifier_largeur(-2)
except ValueError:
    print("Entrée invalide repérée")
```

### Rappel — Une expression exige des mots contigus

« Grande salle » apparaît dans « La grande salle ouvre », mais pas dans « La grande porte mène à la salle ». La présence des deux mots ne prouve pas qu'ils se suivent.

Vous avez découvert `PhraseMatcher` au TD2. Il compare ici les formes en minuscules (`attr="LOWER"`) à un motif tokenisé par `nlp.make_doc(...)`. Il renvoie des indices de **tokens**. Le passage `doc[debut:fin]` est un `Span` ; ses attributs `start_char` et `end_char` donnent les bornes en **caractères** dans le texte original.

**Exemple fourni.** Exécutez la comparaison suivante. Elle conserve le passage et ses positions, puis montre la différence entre une expression contiguë et des mots dispersés. Dans votre recherche, ne réunissez pas fragment, forme, lemme et expression sous un même compte.


```python
from spacy.matcher import PhraseMatcher
regles_demo_td3 = PhraseMatcher(nlp.vocab, attr="LOWER")
regles_demo_td3.add("SALLE", [nlp.make_doc("grande salle")])
for texte_demo_expression in ["La grande salle ouvre.", "La grande porte mène à la salle."]:
    doc_demo_expression = nlp(texte_demo_expression)
    print("Texte :", texte_demo_expression)
    for identifiant_demo, debut_demo, fin_demo in regles_demo_td3(doc_demo_expression):
        passage_demo = doc_demo_expression[debut_demo:fin_demo]
        print(passage_demo.text, passage_demo.start_char, passage_demo.end_char)
```

### Cours — Normaliser en annonçant ce que l'on perd

Une **normalisation typographique** rapproche des graphies selon une règle annoncée : par exemple, ignorer la casse, uniformiser les apostrophes et réduire les espaces successifs à un seul. Elle ne traduit pas un mot et ne remplace pas un terme par un synonyme.

Dans la fonction fournie ci-dessous, `replace` uniformise l'apostrophe, `split()` découpe sur les espaces et `" ".join(...)` recompose la chaîne avec un espace entre les morceaux. `lower()` passe le résultat en minuscules.

**Exemple fourni.** Exécutez-le et observez les transformations. Gardez toujours l'original : les positions dans la copie normalisée ne sont plus nécessairement ses positions.


```python
def normaliser_typographie(texte):
    return " ".join(texte.replace("’", "'").split()).lower()

print(normaliser_typographie(" L’archive   décrit "))
```
