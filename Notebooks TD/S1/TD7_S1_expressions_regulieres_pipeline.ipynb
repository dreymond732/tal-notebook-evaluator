## Découvrir le module `re`

Une expression régulière, souvent appelée **regex**, décrit une forme à rechercher dans une chaîne : un mot, une suite de chiffres ou une date, par exemple. Le module `re` fournit les outils pour chercher, extraire et remplacer ces formes. Il est disponible avec Python : exécutez `import re` dans la cellule suivante, sans installation supplémentaire.

Les fonctions s’appellent avec le préfixe `re.`, suivi de leur nom. Leurs arguments précisent notamment le motif et le texte concernés ; une fonction de remplacement reçoit aussi le texte de remplacement. Vous découvrirez chaque fonction sur un exemple avant de l’utiliser dans un exercice.


```python
import re
```

## Rappel — Motif et texte

Le module `re` cherche des motifs : `search` repère une occurrence, `findall` renvoie toutes les occurrences et `sub` remplace. Une regex décrit une forme ; elle ne réalise pas une analyse linguistique complète.


### Cours — Du motif à la présence

Une expression régulière décrit une forme à rechercher. Une chaîne préfixée par `r`, dite **brute**, garde les antislashs utiles aux motifs. `re.search` renvoie un objet de correspondance, ou `None` si le motif est absent ; `bool(...)` transforme ce résultat en présence/absence. L'option `re.IGNORECASE` ignore la différence entre majuscules et minuscules. `re.match` ne cherche qu'au début de la chaîne ; son comportement diffère donc de `search`.

Un point non échappé (`.`) désigne un caractère quelconque, sauf un saut de ligne avec les options utilisées ici ; `\.` désigne un point littéral. 

**Exemple à observer.** Prédisez les trois booléens, puis comparez les recherches. Le dernier affichage montre la recherche d’un point littéral.


```python
phrase_exemple = "Un MUSÉE ouvre."
print(bool(re.search(r"musée", phrase_exemple, re.IGNORECASE)))
print(bool(re.search(r"bibliothèque", phrase_exemple, re.IGNORECASE)))
print(bool(re.match(r"musée", phrase_exemple, re.IGNORECASE)))
print(re.findall(r"\.", "version 1.2, salle A"))
```

### Cours — Classes et répétitions

Une classe comme `[0-9]` décrit un chiffre. Le quantificateur `+` signifie « une ou plusieurs répétitions » ; `{2}` signifie « exactement deux ». `findall` renvoie une liste de chaînes dans l'ordre des occurrences. Extraire des chiffres ne les convertit pas automatiquement en nombres utilisables dans un calcul.


```python
repere_exemple = "Salle 8, boîte 41 et étagère 205"
print(re.findall(r"[0-9]", repere_exemple))
print(re.findall(r"[0-9]+", repere_exemple))
```

### Cours — Délimiter ce que l’on appelle un mot

`[A-Za-z]` décrit des lettres latines non accentuées. Il ne couvre pas à lui seul les lettres françaises accentuées. Une classe peut être étendue en écrivant les caractères supplémentaires acceptés. Une apostrophe ou un chiffre ne fait pas partie d’une classe composée uniquement de lettres.

Le découpage par regex et celui par espaces ne définissent donc pas la même unité. Une suite de lettres peut constituer seulement un fragment d’un mot avec apostrophe ; inversement, `split()` peut laisser la ponctuation attachée aux mots.

**Exemple à observer.** Comparez les deux classes de lettres, puis le résultat de `split()`. Repérez les caractères qui expliquent les différences.


```python
exemple_accents = "L’été : musée 4, École !"
print(re.findall(r"[A-Za-z]+", exemple_accents))
print(re.findall(r"[A-Za-zÉéèêàùçôîïûüœŒ]+", exemple_accents))
print(exemple_accents.split())
```

Une fonction permet d’appliquer la même suite d’opérations à plusieurs textes, sans recopier les instructions.

**Exemple à observer.** Prédisez la casse du marqueur dans les deux affichages, puis expliquez la différence observée.

### Cours — Relier nettoyage, extraction et comptage

Un **pipeline** est une suite de traitements dont les résultats alimentent l’étape suivante. Nettoyer un texte, extraire ses fragments puis les compter forme un tel enchaînement. Changer une étape peut donc modifier les résultats suivants.

Un dictionnaire de fréquences associe à chaque forme son nombre d’occurrences. La liste conserve toutes les occurrences dans leur ordre ; le dictionnaire les regroupe par forme.

**Exemple à observer.** Le comptage ci-dessous reprend le principe du TD4. Prédisez les effectifs et vérifiez que leur somme correspond au nombre d’éléments de départ.


```python
elements_exemple = ["image", "son", "image"]
compte_exemple = {}
for element in elements_exemple:
    if element in compte_exemple:
        compte_exemple[element] = compte_exemple[element] + 1
    else:
        compte_exemple[element] = 1
print(compte_exemple)
```

### Cours — Séparer extraction de formes et interprétation

Un motif trop permissif peut produire des correspondances vides : `*` signifie « zéro ou plusieurs », alors que `+` en exige au moins une. La longueur de la liste obtenue peut alors inclure des chaînes vides. Même un motif correctement écrit n'indique pas, à lui seul, le sens d'un mot ou sa fonction dans une phrase. Une sortie calculée s’interprète donc en fonction de la convention d’extraction retenue et de ses limites.

**Exemple de distinction entre forme et sens.** Dans « la porte est ouverte » et « elle porte un sac », on retrouve la suite de lettres `porte`. La seule recherche de ces lettres ne distingue pas le nom du verbe : leur emploi dépend de la phrase. Pour votre réponse personnelle, choisissez un autre cas.

**Exemple à observer.** Comparez les deux listes avant de compter leurs éléments : le même texte est testé avec `*`, puis avec `+`.


```python
print(re.findall(r"[0-9]*", "ab"))
print(re.findall(r"[0-9]+", "ab"))
```
