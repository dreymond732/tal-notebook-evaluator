### Premier parcours : une paire à chaque tour — Q4 : 25 min

`for` répète un bloc pour chaque élément. Le signe `:` ouvre le bloc ; les lignes indentées appartiennent au parcours. `items()` fournit les paires clé–valeur d’un dictionnaire. Dans `for cle, valeur in dictionnaire.items():`, les deux noms reçoivent séparément les deux composantes de la paire à chaque tour : c’est le déballage. Créez la liste d’accumulation **avant** la boucle ; sinon elle serait réinitialisée à chaque tour.

Avant d’exécuter l’exemple, prédisez son affichage. Il utilise des données différentes de celles de la question.


```python
couleurs_parcours = {"white": "blanc", "black": "noir"}
fiches_couleurs = []
for cle_couleur, valeur_couleur in couleurs_parcours.items():
    print("Paire courante :", cle_couleur, valeur_couleur)
    fiche_couleur = f"{cle_couleur} : {valeur_couleur}"
    fiches_couleurs.append(fiche_couleur)
print("Liste construite :", fiches_couleurs)
```

### Comparer les présences — Q6 : 14 min

L’intersection `ensemble_a & ensemble_b` conserve les éléments des deux ensembles. La différence `ensemble_a - ensemble_b` conserve ceux de A absents de B. Elle n’est pas symétrique : inverser les opérandes change la question. Les méthodes `.intersection()` et `.difference()` expriment les mêmes opérations.

Avant d’exécuter l’exemple, prédisez son affichage. Il utilise des données différentes de celles de la question.


```python
rayon_a = {"roman", "poésie", "essai"}
rayon_b = {"poésie", "théâtre"}
print("Commun :", rayon_a & rayon_b)
print("Seulement A :", rayon_a - rayon_b)
print("Seulement B :", rayon_b.difference(rayon_a))
```
