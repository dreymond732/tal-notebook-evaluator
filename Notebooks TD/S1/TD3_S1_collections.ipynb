## Cours — Quel type de collection choisir ?

Une **liste** conserve l’ordre des éléments et peut être modifiée : elle convient à une suite de mots à enrichir. Un **tuple** est lui aussi ordonné, mais ses éléments ne peuvent pas être remplacés : il peut représenter une référence stable. Un **dictionnaire** associe chaque clé à une valeur, par exemple un mot à sa traduction. Un **ensemble** ne conserve qu’un exemplaire de chaque valeur : il permet de comparer des vocabulaires sans tenir compte des répétitions.

Vous allez manipuler ces quatre collections, puis justifier leur emploi. Les crochets d’une liste et d’un accès par indice ont été rencontrés au TD2.

## Cours — Premier parcours : une paire à chaque tour — Q4 : 25 min

Une boucle `for` répète des instructions pour chaque élément d’une collection. Chaque répétition est appelée un **tour**. Le signe `:` termine la ligne qui ouvre la boucle. Le bloc répété est **indenté** : ses lignes commencent par quatre espaces. Une ligne revenue au début de la cellule est exécutée après la boucle.

La méthode `items()` fournit les paires clé–valeur d’un dictionnaire. Dans `for cle, valeur in dictionnaire.items():`, `cle` et `valeur` reçoivent les deux composantes de la paire courante. Cette affectation à deux noms s’appelle le **déballage**. Les noms peuvent être choisis pour rendre leur rôle explicite.

Pour construire une liste, on commence par une liste vide, écrite `[]`, **avant** la boucle. À chaque tour, `append()` y ajoute un élément. Si l’on recréait la liste vide dans le bloc, on perdrait les ajouts des tours précédents. Une f-string, vue au TD1, permet ici de transformer chaque paire en une courte fiche.

### Exemple à observer

Prédisez le nombre de tours et de fiches produites. Pendant la lecture du résultat, suivez la paire courante puis l’ajout de sa fiche à la liste. Repérez les trois lignes répétées et l’affichage final, placé après la boucle.


```python
couleurs_parcours = {"white": "blanc", "black": "noir"}
fiches_couleurs = []
for cle_couleur, valeur_couleur in couleurs_parcours.items():
    print("Paire courante :", cle_couleur, valeur_couleur)
    fiche_couleur = f"{cle_couleur} : {valeur_couleur}"
    fiches_couleurs.append(fiche_couleur)
print("Liste construite :", fiches_couleurs)
```

## Cours — Comparer les présences — Q6 : 14 min

L’intersection `ensemble_a & ensemble_b` conserve les éléments présents à la fois dans les deux ensembles. La différence `ensemble_a - ensemble_b` conserve ceux de A absents de B. Elle n’est pas symétrique : inverser les opérandes change la question. Les méthodes `.intersection()` et `.difference()` expriment les mêmes opérations.

### Exemple à observer

Prédisez les trois ensembles. Comparez en particulier les deux différences : changer leur ordre change les éléments recherchés.


```python
rayon_a = {"roman", "poésie", "essai"}
rayon_b = {"poésie", "théâtre"}
print("Commun :", rayon_a & rayon_b)
print("Seulement A :", rayon_a - rayon_b)
print("Seulement B :", rayon_b.difference(rayon_a))
```
