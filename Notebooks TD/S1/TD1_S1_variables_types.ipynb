### Comparer et combiner — Q3 : 18 min

`==` compare, alors que `=` affecte. `>=` signifie « supérieur ou égal », et `>` exclut l’égalité. `<` et `<=` expriment les comparaisons inverses, « inférieur » et « inférieur ou égal ». Une comparaison produit `True` ou `False`. Avec `and`, les deux conditions doivent être vraies. On peut conserver cette réponse dans une variable et l’afficher.

Avant d’exécuter l’exemple, prédisez son affichage. Il utilise des données différentes de celles de la question.


```python
places = 18
inscrits = 16
inscription_possible = inscrits < places and places > 0
print("Exemple :", inscrits == places, inscription_possible)
```
