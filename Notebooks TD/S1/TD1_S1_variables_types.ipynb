## Cours — Variables et valeurs

Une variable associe un nom à une valeur. `=` affecte une valeur : Python évalue ce qui est à droite, puis l’associe au nom écrit à gauche. Une nouvelle affectation remplace la valeur précédente. `==` sert à comparer deux valeurs, comme vous le verrez à l’exercice 3.

Choisissez des noms descriptifs, sans espace, par exemple `nombre_pages`. Un nom peut contenir des lettres, des chiffres et `_`, mais ne commence pas par un chiffre. Respectez les majuscules et les minuscules : `prix` et `Prix` sont deux noms différents. N’utilisez pas les mots réservés du langage, comme `True`, ni les noms des fonctions que vous employez, comme `print`.

Chaque valeur a un type : nombre, chaîne de caractères ou booléen. La fonction `type()` permet de l’observer.

## Cours — Comparer et combiner — Q3 : 18 min

`==` compare, alors que `=` affecte. `>=` signifie « supérieur ou égal », et `>` exclut l’égalité. `<` et `<=` expriment les comparaisons inverses, « inférieur » et « inférieur ou égal ». Une comparaison produit `True` ou `False`. Avec `and`, les deux conditions doivent être vraies. On peut conserver cette réponse dans une variable et l’afficher.

### Exemple à observer

Prédisez les deux booléens. Le premier répond à une égalité ; le second combine deux comparaisons avec `and`.


```python
places = 18
inscrits = 16
inscription_possible = inscrits < places and places > 0
print("Exemple :", inscrits == places, inscription_possible)
```

## Cours — Convertir explicitement — Q5 : 18 min

`int(chaine)` transforme une chaîne représentant un entier en nombre entier. Une addition de nombres calcule une somme ; `+` entre chaînes les assemble : c’est une **concaténation**. Elle conserve les caractères tels quels ; un espace entre deux mots doit donc être présent dans l’une des chaînes ou ajouté entre elles. Mélanger chaîne et entier dans une addition provoque une erreur de type. La conversion doit donc précéder le calcul.

### Exemple à observer

Prédisez les deux résultats de l’addition. Comparez ensuite leur écriture et les types affichés : assembler des caractères et additionner des nombres sont deux opérations différentes.


```python
quantite_saisie = "35"
quantite_numerique = int(quantite_saisie)
print("Exemples :", quantite_saisie + "5", quantite_numerique + 5)
print("Types :", type(quantite_saisie), type(quantite_numerique))
```

## Cours — Décrire une phrase personnelle — Q6 : 22 min

`len(chaine)` compte ses caractères, espaces et ponctuation compris. L’expression `fragment in chaine` vérifie la présence d’une sous-chaîne ; la casse, c’est-à-dire la distinction entre majuscules et minuscules, compte. `repr(chaine)` produit une représentation textuelle avec des guillemets et des signes pour les retours à la ligne. L’afficher avec `print()` aide à repérer ces caractères et à comparer votre phrase aux résultats ; cela ne modifie pas la variable de départ.

### Exemple à observer

Prédisez la longueur et les deux recherches. Comparez les résultats pour « Atelier » et « atelier », puis observez comment `repr()` délimite la chaîne.


```python
annonce_courte = "Atelier du soir !"
print("Exemples :", len(annonce_courte), "Atelier" in annonce_courte, "atelier" in annonce_courte)
print("Texte conservé :", repr(annonce_courte))
```
