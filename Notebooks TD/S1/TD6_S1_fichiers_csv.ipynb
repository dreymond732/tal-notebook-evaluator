### Cours — Dédoublonner avec un ensemble

Un ensemble représente des valeurs distinctes : ajouter deux fois la même valeur ne crée pas deux éléments. `set()` crée un ensemble vide et sa méthode `add(valeur)` y ajoute une valeur. Un dictionnaire peut associer une clé à un ensemble, par exemple pour regrouper les salles utilisées par chaque atelier. L’ordre d’affichage d’un ensemble ne change pas son contenu.

**Exemple à observer.** Suivez les ajouts dans l’atelier `image`. Prédisez combien de salles distinctes son ensemble contiendra après l’ajout répété.


```python
salles_par_atelier = {"image": {"Salle bleue"}, "son": set()}
salles_par_atelier["image"].add("Salle verte")
salles_par_atelier["image"].add("Salle bleue")
print(salles_par_atelier)
print(len(salles_par_atelier["image"]))
```

### Cours — Écrire puis relire un CSV

Un **CSV** est un fichier texte organisé en lignes et en colonnes. Un séparateur délimite les colonnes ; la première ligne peut contenir leurs noms, appelés **en-tête**. Le module `csv`, disponible avec Python, gère cette écriture. `import csv` donne accès à ses fonctions.

`csv.writer(fichier)` prépare un objet qui écrit une ligne avec sa méthode `writerow`. Cette méthode reçoit les valeurs de la ligne, par exemple sous forme de liste. Le séparateur par défaut est la **virgule**. Le mode `"w"` crée ou remplace le fichier ; `newline=""` laisse le module gérer les fins de ligne. Relire le fichier permet d’observer le texte réellement enregistré.

Pour réunir plusieurs chaînes dans une cellule du tableau, `join()` peut recevoir une liste ou un ensemble de chaînes. Avec un ensemble, l’ordre des éléments réunis n’est pas fixé ; cela ne change pas les affiliations présentes.

**Exemple à observer.** Repérez l’ouverture du fichier, l’écriture de l’en-tête et celle des données. Après exécution, comparez chaque ligne relue aux valeurs écrites.


```python
import csv
with open("ateliers_exemple.csv", "w", encoding="utf-8", newline="") as sortie_exemple:
    ecrivain_exemple = csv.writer(sortie_exemple)
    ecrivain_exemple.writerow(["atelier", "capacite"])
    ecrivain_exemple.writerow(["photo", 12])
    ecrivain_exemple.writerow(["son", 8])
with open("ateliers_exemple.csv", "r", encoding="utf-8") as entree_exemple:
    verification_exemple = entree_exemple.read()
print(verification_exemple)
```
