### Atelier préparatoire — Lire un fichier et refermer la ressource

L'exemple crée un fichier **distinct** du fichier d'affiliations. Les outils `Path` de la préparation sont fournis ; vous pratiquez ici la lecture avec `with open`.

### Atelier préparatoire — Limiter le découpage

Dans `split(";", 1)`, le premier argument désigne le séparateur et le second le **nombre maximal de découpages**. Le résultat est une liste de deux fragments au plus. Comparez cet appel au découpage sans limite : un point-virgule peut appartenir à la seconde partie de la donnée. Les espaces autour d'un fragment ne disparaissent pas automatiquement.


```python
ligne_exemple = "atelier image;Salle bleue;annexe"
print(ligne_exemple.split(";"))
print(ligne_exemple.split(";", 1))
```

### Atelier préparatoire — Écrire puis relire un CSV

`csv.writer` prépare un objet qui écrit une ligne avec `writerow`. Le mode `"w"` crée ou remplace le fichier ; `newline=""` laisse le module gérer les fins de ligne. L'en-tête est lui aussi une ligne. Le séparateur de colonnes est ici la **virgule**, convention par défaut de `csv.writer`.

Pour Q6, gardez les colonnes `personne`, `nb_affiliations`, `affiliations`. Dans la dernière colonne, réunissez les affiliations avec `" | "` : ce séparateur interne est distinct de la virgule qui sépare les colonnes. L'ordre des personnes et des affiliations est libre. Relisez le fichier et contrôlez l'en-tête, le nombre de lignes et la correspondance entre un nombre d'affiliations et son contenu.


```python
import csv
with open("ateliers_exemple.csv", "w", encoding="utf-8", newline="") as sortie_exemple:
    ecrivain_exemple = csv.writer(sortie_exemple)
    ecrivain_exemple.writerow(["atelier", "capacite"])
    ecrivain_exemple.writerow(["photo", 12])
    ecrivain_exemple.writerow(["son", 8])
with open("ateliers_exemple.csv", "r", encoding="utf-8") as entree_exemple:
    verification_exemple = entree_exemple.read()
print(verification_exemple)
```
