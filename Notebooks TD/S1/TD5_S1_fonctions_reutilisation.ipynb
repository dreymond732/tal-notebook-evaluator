## Rappel — `print` et `return`

`print` rend un résultat visible. `return` transmet une valeur à la suite du programme. Une fonction doit recevoir ses données par paramètres plutôt que dépendre silencieusement d’une variable extérieure.


### Atelier préparatoire — Définir, appeler et recevoir une valeur

Une définition commence par `def`, suivie du nom de la fonction, de ses paramètres entre parenthèses et de `:`. Les lignes indentées forment son corps. Un **paramètre** reçoit une valeur à chaque appel ; `return` transmet le résultat. La définition seule ne lance pas le calcul. `print` affiche une valeur, mais ne la transmet pas comme résultat de la fonction.

Dans l'exemple, prédisez les affichages et la valeur de `recu`. Expliquez pourquoi la dernière ligne affiche aussi une valeur particulière. Ensuite, réalisez l'essai personnel : ses données sont différentes de celles de l'exercice.


```python
def encadrer(message):
    return "[" + message + "]"

def montrer(message):
    print("[" + message + "]")

recu = encadrer("atelier")
print(recu)
print(montrer("atelier"))
```

### Atelier préparatoire — Réutiliser un résultat et un paramètre par défaut

Composer consiste à transmettre le résultat d'une fonction à une autre, sans recopier son traitement. Dans l'exemple, `etiqueter` appelle `encadrer`, défini plus haut. Le paramètre `categorie="note"` possède une **valeur par défaut** : elle est utilisée uniquement si l'appel ne fournit pas de deuxième argument. Cela prépare les fonctions réutilisables du contrôle ; les fonctions demandées ensuite gardent leurs signatures.

Pour le comptage de mots à venir, rappelez-vous que `split()` sans argument regroupe les espaces successifs et renvoie une liste vide pour une chaîne vide.


```python
def etiqueter(message, categorie="note"):
    return encadrer(categorie + ": " + message)

print(etiqueter("à relire"))
print(etiqueter("validé", "archive"))
print("".split())
```
