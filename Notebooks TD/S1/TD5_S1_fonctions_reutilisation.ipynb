## Rappel — `print` et `return`

`print` rend un résultat visible. `return` transmet une valeur à la suite du programme. Une fonction doit recevoir ses données par paramètres plutôt que dépendre silencieusement d’une variable extérieure.


### Cours — Définir, appeler et recevoir une valeur

Une définition commence par `def`, suivie du nom de la fonction, de ses paramètres entre parenthèses et de `:`. Les lignes indentées forment son corps. Un **paramètre** reçoit une valeur à chaque appel ; `return` transmet le résultat. La définition seule ne lance pas le calcul. Les paramètres et les variables créées dans le corps sont locaux à la fonction : ils servent pendant son exécution. `print` affiche une valeur, mais ne la transmet pas comme résultat de la fonction.

Une fonction sans `return` renvoie la valeur spéciale `None`, qui indique ici l’absence de résultat à réutiliser. Ce n’est pas le texte qu’elle a affiché.

**Exemple à observer.** Prédisez la valeur de `recu` et les affichages. Après exécution, distinguez le texte affiché par `montrer` de la valeur `None` affichée par le `print` qui entoure son appel.


```python
def encadrer(message):
    return "[" + message + "]"

def montrer(message):
    print("[" + message + "]")

recu = encadrer("atelier")
print(recu)
print(montrer("atelier"))
```

### Cours — Réutiliser un résultat et un paramètre par défaut

Composer consiste à transmettre le résultat d'une fonction à une autre, sans recopier son traitement. Dans l'exemple, `etiqueter` appelle `encadrer`, défini plus haut. Le paramètre `categorie="note"` possède une **valeur par défaut** : elle est utilisée uniquement si l'appel ne fournit pas de deuxième argument. Le même appel peut ainsi employer la valeur prévue ou en choisir une autre.

Pour compter les fragments d’une chaîne, `split()` sans argument regroupe les espaces successifs et renvoie une liste vide pour une chaîne vide.


```python
def etiqueter(message, categorie="note"):
    return encadrer(categorie + ": " + message)

print(etiqueter("à relire"))
print(etiqueter("validé", "archive"))
print("".split())
```
