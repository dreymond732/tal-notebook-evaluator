### Comprendre les bornes de range — Q4 : 15 min

`range(debut, fin, pas)` génère des entiers : début inclus, fin exclue. Sans pas explicite, le pas vaut 1. Avec un seul argument, le départ vaut 0 et cet argument donne la fin exclue. Un pas négatif permet de reculer, avec une borne finale plus petite que le début. `list(range(...))` rend les valeurs visibles. La borne exclue doit être choisie pour inclure la dernière valeur souhaitée.

Avant d’exécuter l’exemple, prédisez son affichage. Il utilise des données différentes de celles de la question.

Une autre manière de sélectionner des entiers consiste à tester leur reste de division avec `%` : `nombre % diviseur`. Un entier est pair lorsque le reste de sa division par 2 vaut 0. Cette sélection par condition et le parcours avec un pas de 2 sont deux démarches possibles ; comparez-les sans confondre `%` avec une division.


```python
print("Croissant :", list(range(2, 9, 3)))
print("Décroissant :", list(range(9, 1, -2)))
for position_demo in range(3):
    print("Position :", position_demo)

# Autre donnée : reste d’une division et test de parité.
nombre_demo = 13
print("Reste par 5 :", nombre_demo % 5)
print("Pair :", nombre_demo % 2 == 0)
```

### Accumuler des effectifs dans un dictionnaire — Q6 : 17 min

À chaque occurrence, une fréquence augmente de 1. `.get(cle, 0)` lit la fréquence déjà mémorisée, ou 0 lorsque la clé est absente. Une autre écriture distingue explicitement les deux cas avec `if/else`. Les deux exemples suivants font le même calcul sur des données différentes de Q6 ; aucun ordre particulier des clés n’est demandé.

Avant d’exécuter l’exemple, prédisez son affichage. Il utilise des données différentes de celles de la question.


```python
observations_demo = ["revue", "livre", "revue"]
comptage_demo = {}
for forme_demo in observations_demo:
    comptage_demo[forme_demo] = comptage_demo.get(forme_demo, 0) + 1
print("Avec get :", comptage_demo)
autre_comptage = {}
for forme_demo in observations_demo:
    if forme_demo in autre_comptage:
        autre_comptage[forme_demo] = autre_comptage[forme_demo] + 1
    else:
        autre_comptage[forme_demo] = 1
print("Avec condition :", autre_comptage)
```

### Relire une compréhension de liste — Q7 : 11 min

Une compréhension rassemble une transformation, un parcours et éventuellement un filtre : `[transformation for element in collection if condition]`. Lisez d’abord le parcours, puis le filtre, puis la transformation. Elle ne remplace pas la compréhension du raisonnement : les boucles explicites des premières questions restent demandées.

Avant d’exécuter l’exemple, prédisez son affichage. Il utilise des données différentes de celles de la question.


```python
mesures_demo = [2, 5, 1, 4]
doubles_demo = []
for mesure_demo in mesures_demo:
    if mesure_demo >= 4:
        doubles_demo.append(mesure_demo * 2)
doubles_compacts = [mesure_demo * 2 for mesure_demo in mesures_demo if mesure_demo >= 4]
print("Deux écritures :", doubles_demo, doubles_compacts)
```
