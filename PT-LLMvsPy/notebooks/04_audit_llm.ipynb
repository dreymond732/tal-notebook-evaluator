Le tableau du 02 fournit déjà un point de départ. Le total d’un champ y additionne plusieurs termes. Son détail permet de retrouver le nombre associé à chaque entrée. Pour les auteurs, une même mention peut prendre plusieurs formes.

```python
ENQUETE = "polysémie"  # Autres choix : "modalisation" ou "polyphonie"
expressions = {
    "polysémie": ["droit", "état"],
    "modalisation": ["je crois", "je ne crois pas", "il faut", "il faudrait"],
    "polyphonie": ["Turgeon", "Joran"],
}[ENQUETE]
if ENQUETE == "polysémie":
    champ = next(ligne for ligne in lexique["champs"] if ligne["champ"] == "droit_institutions")
    print("02 · Champ droit et institutions : ", champ["formes"], "formes,", champ["lemmes"], "lemmes comptés")
    if lexique["champ_choisi"] == champ["champ"]:
        for terme in expressions:
            detail = lexique["detail_champ"]
            print(terme, "· formes :", detail["formes"][terme], "· lemmes :", detail["lemmes"][terme])
elif ENQUETE == "modalisation":
    print("02 ·", {s: lexique["locutions"][s.replace(" ", "_")]["occurrences"] for s in expressions})
elif ENQUETE == "polyphonie":
    for personne in ["Charles Turgeon", "Théodore Joran"]:
        print("02 ·", personne, lexique["onomastique"][personne]["variantes"])

contextes = {s: concordances_expression(texte, s, largeur=130) for s in expressions}
print("Nombre de correspondances pour chaque chaîne :", {s: len(lignes) for s, lignes in contextes.items()})
exemples = []
for forme, lignes in contextes.items():
    if lignes:
        for i in sorted({0, len(lignes)//2, len(lignes)-1}):
            ligne = lignes[i]
            exemples.append({"forme": forme, **ligne})
for i, ligne in enumerate(exemples):
    print(f'{i} · {ligne["gauche"]}[{ligne["expression"]}]{ligne["droite"]}')
```

```python
INDICES = [0, 1]  # Modifiez ces numéros d’après votre liste.
passages_retenus = []
for i in dict.fromkeys(INDICES):
    if 0 <= i < len(exemples):
        ligne = exemples[i]
        a, b = ligne["debut"], ligne["fin"]
        debut, fin = max(0, a-300), min(len(texte), b+300)
        passages_retenus.append({"forme": ligne["forme"], "debut": debut, "fin": fin,
                                 "debut_occurrence": a, "fin_occurrence": b,
                                 "texte": texte[debut:fin]})
        print(f"Extrait {i}, fenêtre [{debut}:{fin}]", texte[debut:fin])
    else:
        print("Numéro absent de la liste :", i)
LECTURE_1 = ""  # Sens ou voix et indice précis du premier passage.
LECTURE_2 = ""  # Sens ou voix et indice précis du second passage.
```
