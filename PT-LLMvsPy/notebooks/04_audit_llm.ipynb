## 3. Retrouver les mots d’une citation et identifier la voix

Une concordance aligne une occurrence avec quelques caractères à sa gauche et à sa droite. Ce voisinage, le **cotexte**, aide à identifier le sens d’un mot et la voix qui s’exprime. La largeur règle le nombre de caractères affichés de chaque côté.

### Choisir ce qu’on recherche

`concordances_exactes(texte, motif)` retrouve aussi le motif **à l’intérieur** d’un mot, en respectant les majuscules et les minuscules. `concordances_mot(texte, mot, largeur=45)` recherche le **mot entier**, en ignorant la casse. Cette seconde fonction conserve la graphie trouvée sous `pivot` et les positions dans le texte original. Les variantes fléchies ont leur propre graphie.

**Votre exemple.** Avec *droit*, écrivez une phrase contenant une majuscule, un mot plus long tel qu’*endroit* et un pluriel. Prévoyez les trois effectifs du tableau. Entre les deux premières colonnes, seule la casse change. Entre les deux dernières, seule la frontière du mot change.

````python
mon_texte_mot = "À cet endroit, Léa distingue Droit, droits et droit."
mon_mot = "droit"
prevision_mot = ""  # Mes trois nombres et les graphies que j’exclus.

import re

def comparer_recherches(source, mot):
    motifs = list(re.finditer(re.escape(mot), source, flags=re.IGNORECASE))
    entiers = concordances_mot(source, mot)
    debuts_entiers = {ligne["debut"] for ligne in entiers}
    rejet = next((m for m in motifs if m.start() not in debuts_entiers), None)
    if rejet:
        a, b = rejet.span()
        print("Sous-chaîne écartée, en contexte :",
              source[max(0, a-20):a] + "[" + source[a:b] + "]" + source[b:b+20])
    return {"sous-chaîne, casse exacte": len(concordances_exactes(source, mot)),
            "sous-chaîne, casse ignorée": len(motifs), "mot entier, casse ignorée": len(entiers)}

comptes_essai_mot = comparer_recherches(mon_texte_mot, mon_mot)
display(pd.DataFrame([comptes_essai_mot]))
for ligne in concordances_mot(mon_texte_mot, mon_mot):
    print(ligne["debut"], ligne["fin"], ligne["gauche"] + "[" + ligne["pivot"] + "]" + ligne["droite"])
````

### Vérifier une citation avant de l’attribuer

`verifier_citation(texte, citation)` renvoie `EXACTE` quand la chaîne est retrouvée telle quelle ou `NORMALISEE` après normalisation de la casse, des apostrophes et des espaces. `NON_RETROUVEE` décrit l’issue de ces deux recherches. Vous pouvez alors examiner une piste avec `concordances_expression(texte, expression, largeur=...)`, qui retrouve une suite de mots entiers, sans distinguer la casse. Le statut `CANDIDAT_APPROCHE` désigne un passage que **vous** avez délimité pour l’examiner.

**Votre exemple.** Composez deux phrases où une personne rapporte un avis puis le discute. Choisissez une citation et une expression à localiser.
````python
contextes = {}
for expression in expressions:
    if ENQUETE == "modalisation":
        lignes = concordances_expression(texte, expression, largeur=130)
    else:
        lignes = concordances_mot(texte, expression, largeur=130)
    contextes[expression] = lignes
print("Nombre de correspondances pour chaque chaîne :", {s: len(lignes) for s, lignes in contextes.items()})
exemples = []
for forme, lignes in contextes.items():
    if lignes:
        for i in sorted({0, len(lignes)//2, len(lignes)-1}):
            ligne = lignes[i]
            exemples.append({"forme": forme, **ligne})
for i, ligne in enumerate(exemples):
    a, b = ligne["debut"], ligne["fin"]
    print(f'{i} · {ligne["gauche"]}[{texte[a:b]}]{ligne["droite"]}')
````
