### Calculer un TTR et modifier une règle d’intervalle

Comptez les mots de vos phrases ($N$) et leurs formes différentes ($V$), puis calculez le TTR ($V/N$). Un TTR de 0,5 vaut 50 %. Prévoyez si votre TTR tombera entre les deux bornes choisies pour cet essai. Vérifiez votre fraction et comparez les verdicts de la bibliothèque selon que les deux personnes ont suivi la même convention ou des conventions différentes.

````python
bornes_ttr = [0.50, 0.75]  # Modifiez les deux bornes, de la plus petite à la plus grande.
prevision_ratio = ""  # Ma fraction V/N, sa valeur et le verdict prévu.
mots_essai = [mot for phrase in phrases_mots for mot in phrase]
if not mots_essai:
    raise ValueError("Écrivez au moins un mot alphabétique dans vos phrases.")
v_essai, n_essai = len(set(mots_essai)), len(mots_essai)
ttr_essai = v_essai / n_essai
fiche_ratio = {"unite": "formes distinctes / occurrences alphabétiques", "fenetre": "mes phrases",
               "normalisation": "formes minuscules", "agregation": "V/N", "intervalle": bornes_ttr}
verdict_essai_ratio = evaluer_affirmation(fiche_ratio, ttr_essai, protocole_comparable=True)
print("V, N, TTR, intervalle :", v_essai, n_essai, ttr_essai, bornes_ttr)
print("Mêmes conventions :", verdict_essai_ratio)
print("Conventions différentes :", evaluer_affirmation(fiche_ratio, ttr_essai, protocole_comparable=False))
````
